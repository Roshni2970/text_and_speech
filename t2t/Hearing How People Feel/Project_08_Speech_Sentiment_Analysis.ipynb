# Speech Sentiment Analysis

**College Mini Project – Text and Speech Analysis**

Convert speech to text and perform simple sentiment analysis on the recognized text.


## Setup

This project uses `SpeechRecognition`. In Google Colab, microphone access may require a browser-compatible recording approach. The notebook code below is designed to be simple and easy to adapt.


```python

# Install once if needed:
# !pip install SpeechRecognition pyaudio

import speech_recognition as sr

recognizer = sr.Recognizer()

def recognize_speech():
    with sr.Microphone() as source:
        print("Speak now...")
        recognizer.adjust_for_ambient_noise(source, duration=1)
        audio = recognizer.listen(source)
    try:
        text = recognizer.recognize_google(audio)
        print("Recognized text:", text)
        return text
    except sr.UnknownValueError:
        print("Could not understand the speech.")
        return ""
    except sr.RequestError as e:
        print("Speech service error:", e)
        return ""
```

## Analyze Speech Sentiment


```python

import re
from collections import Counter

STOP_WORDS = {
    "the","is","a","an","and","or","of","to","in","on","for","with","this","that",
    "was","are","be","as","at","by","it","from","we","you","he","she","they","i"
}

POSITIVE_WORDS = {
    "good","great","excellent","happy","love","like","amazing","wonderful",
    "best","nice","success","awesome","positive","enjoy"
}
NEGATIVE_WORDS = {
    "bad","poor","sad","hate","dislike","terrible","worst","awful",
    "negative","angry","problem","fail","failure"
}

def clean_words(text):
    return re.findall(r"[A-Za-z']+", text.lower())

def sentiment(text):
    words = clean_words(text)
    score = sum(w in POSITIVE_WORDS for w in words) - sum(w in NEGATIVE_WORDS for w in words)
    label = "Positive" if score > 0 else "Negative" if score < 0 else "Neutral"
    return score, label

text = recognize_speech()
if text:
    score, label = sentiment(text)
    print("Sentiment:", label)
    print("Score:", score)
```
