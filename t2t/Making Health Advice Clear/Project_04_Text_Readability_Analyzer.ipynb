# Text Readability Analyzer

**College Mini Project – Text and Speech Analysis**

Estimate the readability of a text using simple sentence, word, and syllable-based calculations.


## Enter Text


```python
import re

def syllables(word):
    word = re.sub(r"[^a-z]", "", word.lower())
    groups = re.findall(r"[aeiouy]+", word)
    count = len(groups)
    return max(1, count)

text = input("Enter text: ")
words = re.findall(r"\\b[A-Za-z]+\\b", text)
sentences = max(1, len([s for s in re.split(r"[.!?]+", text) if s.strip()]))
syllable_count = sum(syllables(w) for w in words)
score = 206.835 - 1.015 * (len(words)/sentences) - 84.6 * (syllable_count/max(1,len(words)))
print("Words:", len(words))
print("Sentences:", sentences)
print("Estimated readability score:", round(score, 2))
```
