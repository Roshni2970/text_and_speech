# Keyword Extractor

**College Mini Project – Text and Speech Analysis**

Extract important keywords from a paragraph by removing common stop words.


## Enter Text


```python

import re
from collections import Counter

STOP_WORDS = {
    "the","is","a","an","and","or","of","to","in","on","for","with","this","that",
    "was","are","be","as","at","by","it","from","we","you","he","she","they","i"
}

POSITIVE_WORDS = {
    "good","great","excellent","happy","love","like","amazing","wonderful",
    "best","nice","success","awesome","positive","enjoy"
}
NEGATIVE_WORDS = {
    "bad","poor","sad","hate","dislike","terrible","worst","awful",
    "negative","angry","problem","fail","failure"
}

def clean_words(text):
    return re.findall(r"[A-Za-z']+", text.lower())

def sentiment(text):
    words = clean_words(text)
    score = sum(w in POSITIVE_WORDS for w in words) - sum(w in NEGATIVE_WORDS for w in words)
    label = "Positive" if score > 0 else "Negative" if score < 0 else "Neutral"
    return score, label

text = input("Enter a paragraph: ")
words = clean_words(text)
keywords = [w for w in words if w not in STOP_WORDS and len(w) > 2]
print("Keywords:", Counter(keywords).most_common(10))
```
