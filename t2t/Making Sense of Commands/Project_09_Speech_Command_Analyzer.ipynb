# Speech Command Analyzer

**College Mini Project – Text and Speech Analysis**

Recognize simple spoken commands such as hello, start, stop, and exit.


## Setup

This project uses `SpeechRecognition`. In Google Colab, microphone access may require a browser-compatible recording approach. The notebook code below is designed to be simple and easy to adapt.


```python

# Install once if needed:
# !pip install SpeechRecognition pyaudio

import speech_recognition as sr

recognizer = sr.Recognizer()

def recognize_speech():
    with sr.Microphone() as source:
        print("Speak now...")
        recognizer.adjust_for_ambient_noise(source, duration=1)
        audio = recognizer.listen(source)
    try:
        text = recognizer.recognize_google(audio)
        print("Recognized text:", text)
        return text
    except sr.UnknownValueError:
        print("Could not understand the speech.")
        return ""
    except sr.RequestError as e:
        print("Speech service error:", e)
        return ""
```

## Recognize Command


```python
text = recognize_speech().lower()
commands = {
    "hello": "Greeting command detected",
    "start": "Start command detected",
    "stop": "Stop command detected",
    "exit": "Exit command detected"
}
matched = next((message for command, message in commands.items() if command in text), None)
print(matched or "No supported command detected.")
```
